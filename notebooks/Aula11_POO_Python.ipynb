## O método especial __init__ (construtor)

Como sabemos que todos os carros que queremos representar possuem as propriedades “cor”, “ano”, e “modelo”, podemos deixar essas propriedades como atributos da classe e inicializá-las quando um objeto é instanciado (criado). 

Para isso, vamos utilizar o método especial __init__, conhecido como construtor da classe.



````python
class Carro:
    def __init__(self, m, a, c):
        self.modelo = m
        self.ano    = a
        self.cor    = c

brasilia = Carro('Brasilia', 1968, 'amarela')

class Fusca(Carro):
    pass
   
fusca_padrao = Fusca('Fusca', 1981, 'preto')

novo_fusca = Fusca('Novo_fusca',1981,'azul')

novo_fusca.ano += 10   # observe que podemos acessar atributos de fusca

print(fusca_padrao.ano)

print(novo_fusca.ano)
````

Observe que o construtor permite criar objetos já com seus atributos inicializados, evitando a necessidade de atribuir cada característica separadamente depois da criação.

````python
class Viga:
    def __init__(self, comprimento, largura, altura):
        self.comprimento = comprimento
        self.largura = largura
        self.altura = altura

viga_1 = Viga(5.0, 0.20, 0.50)
viga_2 = Viga(4.0, 0.15, 0.40)

print(viga_1.comprimento)
print(viga_2.comprimento)
````

## Mas o que é esse self?

Todo método de uma classe recebe como primeiro parâmetro uma referência à instância que chama o método, permitindo assim que o objeto acesse os seus próprios atributos e métodos. 

Por convenção, chamamos esse primeiro parâmetro de self, mas qualquer outro nome poderia ser utilizado.

O método especial __init__ é chamado automaticamente após a criação de um objeto, e no caso de um Carro, na nova instância são criados os atributos ‘modelo’, ‘ano’ e ‘cor’ com os valores dados como argumentos do construtor (no caso Carro(“Brasilia”, 1968, “amarela”)).

Assim, no corpo de qualquer método, um atributo pode ser criado, acessado ou modificado usando self.nome_do_atributo.


Para exemplificar a criação de outros métodos e entender melhor o papel do self vamos criar os métodos imprima, acelere e pare. Por ser um exemplo mais elaborado, vamos também colocar os testes dentro da função main.

````python
def main():
    brasilia = Carro('brasilia', 1968, 'amarela', 80)
    fuscao = Carro('fuscao', 1981, 'preto', 95)

#Na função main, brasilia e fuscao são instâncias de Carro. Para chamar o método acelere de Carro, usamos a notação:
 #   Carro.acelere(brasilia, 40)    
 #    Carro.acelere(fuscao, 50)
 #   Carro.acelere(brasilia, 80)
    
 #   Carro.pare(brasilia)
    Carro.acelere(fuscao, 100)

    
class Carro:
    def __init__(self, m, a, c, vm):
        self.modelo = m
        self.ano    = a
        self.cor    = c
        self.vel    = 0
        self.maxV   = vm  # velocidade maxima

    def imprima(self):
        if self.vel == 0: # parado da para ver o ano
            print( "%s %s %d"%(self.modelo, self.cor, self.ano)     )
        elif self.vel < self.maxV:
            print( "%s %s indo a %d Km/h"%(self.modelo, self.cor, self.vel) )
        else:
            print( "%s %s indo muito raaaaaapiiiiiiiidoooooo!"%(self.modelo, self.cor))

    def acelere(self, v):
        self.vel = v
        if self.vel > self.maxV:
            self.vel = self.maxV
        Carro.imprima(self)

    def pare(self):
        self.vel = 0
        Carro.imprima(self)

main()

````

Vamos implementar cálculos geométricos de uma viga. 

Observação: o peso específico de 25 kN/m³ é uma hipótese simplificada para concreto armado. O cálculo da área de forma considera as duas faces laterais e a face inferior, sem considerar as extremidades.

````python
class Viga:
    def __init__(self, comprimento, largura, altura):
        self.comprimento = comprimento
        self.largura = largura
        self.altura = altura

    def calcular_volume(self):
        return self.comprimento * self.largura * self.altura

    def calcular_area_forma(self):
        return 2 * self.comprimento * self.altura + \
               2 * self.largura * self.altura

    def calcular_peso_proprio(self, peso_especifico=25):
        return self.calcular_volume() * peso_especifico


def main():
    viga = Viga(5.0, 0.20, 0.50)

    print("Volume:", viga.calcular_volume(), "m³")
    print("Área de forma:", viga.calcular_area_forma(), "m²")
    print("Peso próprio:", viga.calcular_peso_proprio(), "kN")


main()
````

Essa notação explicita a passagem dos objetos como primeiro parâmetro em cada método, mas é redundante visto que todo objeto sabe a que classe ele pertence. Uma notação mais comum e enxuta é chamar os métodos usando a notação com . de forma semelhante aos atributos e, como o primeiro argumento é sempre ele mesmo, ele pode ser evitado. Assim a chamada:

`Carro.acelere.(brasilia, 40)`

pode ser escrita como:

`brasilia.acelere(40)`



````python
def main():
    brasilia = Carro('brasilia', 1968, 'amarela', 80)
    fuscao = Carro('fuscao', 1981, 'preto', 95)

    brasilia.acelere(40)
    fuscao.acelere(50)
    brasilia.acelere(80)
    brasilia.pare()
    fuscao.acelere(100)

class Carro:
    def __init__(self, m, a, c, vm):
        self.modelo = m
        self.ano    = a
        self.cor    = c
        self.vel    = 0
        self.maxV   = vm  # velocidade maxima

    def imprima(self):
        if self.vel == 0: # parado da para ver o ano
            print( "%s %s %d"%(self.modelo, self.cor, self.ano)     )
        elif self.vel < self.maxV:
            print( "%s %s indo a %d Km/h"%(self.modelo, self.cor, self.vel) )
        else:
            print( "%s %s indo muito raaaaaapiiiiiiiidoooooo!"%(self.modelo, self.cor))

    def acelere(self, v):
        self.vel = v
        if self.vel > self.maxV:
            self.vel = self.maxV
        self.imprima()

    def pare(self):
        self.vel = 0
        self.imprima()

main()

````

## O método especial __str__

O método imprima é muito útil para qualquer tipo de objeto mas não é a forma mais comum utilizando objetos. Em muitos casos, é desejável simplesmente utilizar a função print para imprimir uma “representação textual” do objeto.

Na verdade, quando executamos o comando print(6), como o 6 é um objeto da classe int, a função print converte o inteiro 6 no string 6 antes de ser impresso, algo como print( str(6) ). Lembre-se que, dentre as funções para conversão de tipos (lembra do int em int(input(“Digite um número: ”))?), a função str converte o inteiro 6 para o string 6.

Para exibir então o conteúdo de um objeto usando print, devemos definir o método especial __str__, e devolver um string com a representação textual do objeto. A função str também chama e retorna o valor desse método (se existir).

O código abaixo é basicamente o mesmo que o anterior, mas substituimos o método imprima pelo método especial __str__. No corpo desse método, ao invés de chamar a função print, carrega-se um string apropriado em s, que é retornado pelo método para a função print ou str. Veja que nos métodos acelere e pare, basta chamar print(self).

````python
def main():
    brasilia = Carro('brasilia', 1968, 'amarela', 80)
    fuscao = Carro('fuscao', 1981, 'preto', 95)

    print(brasilia)
    print(fuscao)

    brasilia.acelere(40)
    fuscao.acelere(50)
    brasilia.acelere(80)
    brasilia.pare()
    fuscao.acelere(100)

class Carro:
    def __init__(self, m, a, c, vm):
        self.modelo = m
        self.ano    = a
        self.cor    = c
        self.vel    = 0
        self.maxV   = vm  # velocidade maxima

    def __str__(self):
        if self.vel == 0: # parado da para ver o ano
            s = "%s %s %d"%(self.modelo, self.cor, self.ano)
        elif self.vel < self.maxV:
            s = "%s %s indo a %d Km/h"%(self.modelo, self.cor, self.vel)
        else:
            s = "%s %s indo muito raaaaaapiiiiiiiidoooooo!"%(self.modelo, self.cor)
        return s

    def acelere(self, v):
        self.vel = v
        if self.vel > self.maxV:
            self.vel = self.maxV
        print(self)

    def pare(self):
        self.vel = 0
        print(self)

main()

````

## Importação de classes

Salvar classes em arquivos separados e importá-las em outros programas é uma prática comum em programação. Essa abordagem é usada para organizar o código, facilitar a reutilização e melhorar a manutenção. A seguir, explica-se como isso funciona, por que é vantajoso e como implementá-lo, usando Python como exemplo.

---

### **O que significa salvar classes em arquivos separados?**
Em vez de definir todas as classes e funcionalidades de um programa em um único arquivo, você pode dividir seu código em vários arquivos, cada um contendo classes ou funções relacionadas. Esses arquivos são chamados de **módulos** e podem ser importados em outros programas.

---

### **Por que usar essa abordagem?**
1. **Organização**: O código fica mais limpo e fácil de entender quando funções e classes relacionadas são separadas em arquivos diferentes.
2. **Reutilização**: Classes e funções podem ser reaproveitadas em outros projetos ou partes do mesmo programa, sem a necessidade de reescrever o código.
3. **Manutenção**: É mais fácil encontrar e corrigir bugs ou adicionar funcionalidades quando o código está dividido em partes lógicas.
4. **Colaboração**: Em projetos maiores, diferentes membros da equipe podem trabalhar em arquivos diferentes sem conflitos.
5. **Modularidade**: Facilita a substituição ou atualização de partes específicas do código sem impactar o restante.

---

### **Como salvar e importar classes?**
1. **Criar um arquivo com a classe**  
   Salve sua classe em um arquivo separado. Por exemplo, em um diretório chamado módulos, vamos salvar:

   **`Carro.py`**:
   ```python

        self.ar_condicionado_ligado = True

   ```

2. **Importar a classe em outro arquivo**

---

### **Outras formas de importação**
- Importar várias classes de um arquivo:
  ```python
  from meu_modulo import Classe1, Classe2
  ```

- Importar o módulo inteiro:
  ```python
  import meu_modulo

  obj = meu_modulo.Classe1()
  ```

- Usar apelidos para simplificar:
  ```python
  import meu_modulo as mm

  obj = mm.Classe1()
  ```

---

### **Por que é uma boa prática?**
- **Evita duplicação de código**: Você não precisa recriar a mesma classe em diferentes partes do projeto.
- **Melhora a legibilidade**: Programas grandes tornam-se mais fáceis de entender.
- **Facilita a manutenção**: Alterações em uma classe não exigem edições em múltiplos arquivos.
- **Promove testes unitários**: Cada classe ou módulo pode ser testado separadamente.

---

Essa abordagem modular é um dos princípios fundamentais da programação orientada a objetos e do desenvolvimento de software escalável.

## Importação de classes

Em projetos de Engenharia Civil, um programa pode precisar representar diferentes elementos estruturais, como vigas e pilares. À medida que o programa cresce, manter todas as classes em um único arquivo pode dificultar a organização e a manutenção do código.

Uma alternativa é separar as classes em arquivos diferentes. Em Python, esses arquivos são chamados de **módulos**.

Essa organização permite:

1. **Organização:** separar classes e funcionalidades relacionadas.
2. **Reutilização:** utilizar as mesmas classes em diferentes programas.
3. **Manutenção:** facilitar a correção e atualização do código.
4. **Colaboração:** permitir que diferentes integrantes trabalhem em módulos específicos.

Vamos organizar as classes `ElementoEstrutural`, `Viga` e `Pilar` em um módulo e utilizá-las em um programa principal.

### 1. Organização dos arquivos

Crie uma pasta chamada `projeto_estrutural`, com a seguinte estrutura:

projeto_estrutural/
│
├── modulos/
│   ├── __init__.py
│   └── elementos.py
│
└── main.py

O arquivo `elementos.py` armazenará as classes utilizadas no projeto.

O arquivo `main.py` será responsável por importar essas classes, criar os objetos e executar o programa.

### 2. Criando o módulo elementos.py

No arquivo `elementos.py`, vamos definir uma classe geral chamada `ElementoEstrutural` e duas classes especializadas: `Viga` e `Pilar`.

As classes `Viga` e `Pilar` herdarão os atributos da classe `ElementoEstrutural` e terão características próprias.

````python
class ElementoEstrutural:
    def __init__(self, identificacao, material):
        self.identificacao = identificacao
        self.material = material

    def descrever(self):
        return f"{self.identificacao} - Material: {self.material}"


class Viga(ElementoEstrutural):
    def __init__(self, identificacao, material,
                 comprimento, largura, altura):
        super().__init__(identificacao, material)
        self.comprimento = comprimento
        self.largura = largura
        self.altura = altura

    def calcular_volume(self):
        return self.comprimento * self.largura * self.altura


class Pilar(ElementoEstrutural):
    def __init__(self, identificacao, material,
                 largura, profundidade, altura):
        super().__init__(identificacao, material)
        self.largura = largura
        self.profundidade = profundidade
        self.altura = altura

    def calcular_volume(self):
        return self.largura * self.profundidade * self.altura
````

### 3. Importando as classes no programa principal

Agora, no arquivo `main.py`, vamos importar as classes definidas no módulo `elementos.py`.

A instrução `from ... import ...` permite selecionar quais classes desejamos utilizar no programa.

````python
from modulos.elementos import Viga, Pilar


def main():
    # Criando objetos
    viga = Viga("V01", "Concreto", 5.0, 0.20, 0.50)
    pilar = Pilar("P01", "Concreto", 0.30, 0.30, 3.0)

    # Apresentando os resultados
    print(viga.descrever())
    print("Volume da viga:", viga.calcular_volume(), "m³")

    print()

    print(pilar.descrever())
    print("Volume do pilar:", pilar.calcular_volume(), "m³")


if __name__ == "__main__":
    main()
````

### 4. Outras formas de importação

O Python permite diferentes formas de importar módulos e classes.

**Importar várias classes de um módulo:**

    from modulos.elementos import Viga, Pilar

**Importar o módulo inteiro:**

    import modulos.elementos

    viga = modulos.elementos.Viga(
        "V01", "Concreto", 5.0, 0.20, 0.50
    )

**Importar utilizando um apelido:**

    import modulos.elementos as elementos

    viga = elementos.Viga(
        "V01", "Concreto", 5.0, 0.20, 0.50
    )

A modularização facilita o desenvolvimento de programas maiores, permitindo que as classes sejam organizadas e reutilizadas em diferentes aplicações de Engenharia Civil.

## Síntese final — A importância da Programação Orientada a Objetos

A Programação Orientada a Objetos (POO) é um paradigma de programação que permite organizar sistemas computacionais a partir da representação de entidades do mundo real ou de conceitos abstratos. Em vez de estruturar um programa apenas como uma sequência de instruções e funções, a POO organiza os dados e as operações relacionadas em **classes e objetos**.

Ao longo desta aula, vimos que uma classe define as características e os comportamentos de uma entidade, enquanto os objetos são instâncias dessa classe. Também estudamos atributos, métodos, construtores, referências a objetos, herança, representação textual e organização do código em módulos.

### Quais são as principais vantagens da POO?

- **Organização:** reúne dados e operações relacionadas em estruturas bem definidas, facilitando a compreensão do programa.
- **Reutilização:** permite aproveitar classes e funcionalidades já desenvolvidas, reduzindo a repetição de código.
- **Manutenção:** facilita a identificação de problemas, a correção de erros e a evolução dos sistemas.
- **Abstração:** permite representar apenas as características relevantes de uma entidade, sem precisar reproduzir todos os seus detalhes.
- **Encapsulamento:** organiza e controla o acesso aos dados e às operações de um objeto.
- **Herança:** permite criar novas classes a partir de classes existentes, aproveitando características e comportamentos comuns.
- **Polimorfismo:** permite que diferentes objetos respondam a uma mesma operação de maneiras específicas.

Essas características tornam a POO especialmente útil no desenvolvimento de sistemas complexos, nos quais diferentes entidades possuem informações, propriedades e comportamentos próprios.

### A influência da POO nas aplicações de Engenharia

A POO não está restrita ao desenvolvimento de programas didáticos. Seus conceitos estão presentes na arquitetura de muitos softwares utilizados na Engenharia e em outras áreas do conhecimento.

**1. BIM — Building Information Modeling**

O BIM trabalha com a representação digital de edificações e seus componentes como objetos que possuem geometria, propriedades e informações associadas.

Uma viga, por exemplo, pode ser representada por um objeto com dimensões, material, classificação, propriedades físicas e relações com outros elementos da edificação.

Essa abordagem se aproxima dos princípios da POO: representar entidades por meio de objetos que reúnem características e informações. Embora BIM e POO não sejam a mesma coisa, o paradigma orientado a objetos influencia a estrutura de dados, os modelos de informação e as interfaces de programação utilizadas em aplicações BIM.

**2. Softwares de CAD e modelagem 3D**

Programas de desenho e modelagem computacional trabalham com entidades como linhas, superfícies, sólidos, componentes e montagens. Essas entidades possuem propriedades e operações que podem ser representadas e manipuladas por estruturas orientadas a objetos.

**3. Softwares de análise e simulação estrutural**

Em programas de análise estrutural, elementos como nós, barras, vigas, pilares e materiais podem ser representados computacionalmente por objetos. Cada elemento pode possuir propriedades próprias e métodos relacionados à montagem do modelo e à realização de cálculos.

**4. Sistemas de informação geográfica (SIG/GIS)**

Aplicações de geoprocessamento trabalham com objetos geográficos, como pontos, linhas e polígonos, associados a atributos e informações espaciais. Esses elementos podem ser organizados e manipulados por estruturas orientadas a objetos.

**5. Automação, IoT e sistemas embarcados**

Na automação, sensores, atuadores, controladores e equipamentos podem ser representados como objetos, reunindo seus estados, configurações e operações. Essa organização facilita o desenvolvimento de sistemas que integram hardware e software.

### Conclusão

A Programação Orientada a Objetos oferece uma forma de pensar e estruturar soluções computacionais. Seu valor não está apenas na sintaxe de uma linguagem como Python, mas na capacidade de representar problemas complexos por meio de entidades organizadas, com características e comportamentos definidos.

Na Engenharia, essa forma de organização está relacionada ao desenvolvimento de aplicações que representam edificações, estruturas, materiais, equipamentos, processos e sistemas de automação.

**Portanto, aprender POO significa desenvolver uma capacidade importante para a Engenharia contemporânea: transformar entidades, propriedades, relações e processos de um problema real em modelos computacionais que possam ser manipulados, analisados, simulados e integrados a outros sistemas.**
